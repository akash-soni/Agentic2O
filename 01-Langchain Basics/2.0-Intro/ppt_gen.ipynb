```python
from langchain_openai import ChatOpenAI
from langchain.prompts import ChatPromptTemplate

llm=ChatOpenAI(model="gpt-4o")

print(llm)


```

```python
# Prompt template
prompt_template = ChatPromptTemplate.from_template(
    "Create a PowerPoint slide for a Class 11 Accountancy lesson. Give a title and detailed bullet points on: {topic}"
)
```

```python
# Define slide titles and contents
topics = [
    ("Introduction to Accounting", "Class 11 – Accountancy\nPresented by: Your Name\nSchool Name"),
    ("What is Accounting?", "Accounting is the process of identifying, recording, classifying, summarizing, interpreting, and communicating financial information.\n\nPurpose: To provide financial information to stakeholders for decision-making."),
    ("Objectives of Accounting", "1. To maintain systematic records\n2. To ascertain profit or loss\n3. To determine financial position\n4. To provide accounting information to users"),
    ("Features of Accounting", "• Financial in nature\n• Based on evidence (vouchers, bills)\n• Follows accounting principles\n• Expressed in monetary terms\n• Communicates results to users"),
    ("Functions of Accounting", "1. Measurement\n2. Forecasting\n3. Comparison and Evaluation\n4. Control\n5. Decision-making assistance"),
    ("Process of Accounting", "1. Identifying transactions\n2. Recording in journal\n3. Classifying in ledger\n4. Summarizing (Trial balance & final accounts)\n5. Analyzing & Interpreting\n6. Communicating results"),
    ("Branches of Accounting", "• Financial Accounting\n• Cost Accounting\n• Management Accounting"),
    ("Users of Accounting Information", "• Internal Users: Management, Owners\n• External Users: Investors, Creditors, Government, Tax Authorities"),
    ("Advantages of Accounting", "✅ Provides financial information\n✅ Helps in decision-making\n✅ Assists in tax matters\n✅ Legal evidence\n✅ Facilitates loans"),
    ("Limitations of Accounting", "❌ Ignores qualitative factors\n❌ Based on historical cost\n❌ Possible manipulation\n❌ Does not reflect inflation\n❌ Incomplete without non-financial data"),
    ("Basic Accounting Terms", "• Assets\n• Liabilities\n• Capital\n• Revenue\n• Expenses\n• Profit & Loss\n• Drawings"),
    ("Summary", "• Accounting helps track and manage business finances\n• It serves multiple stakeholders\n• Has both strengths and limitations\n• Foundation for all commerce streams"),
    ("Thank You!", "Any Questions?\n\n“Accounting is the language of business.” – Warren Buffett")
]
```

```python
# Function to add slide
def add_slide(title, content):
    slide_layout = prs.slide_layouts[1]  # Title and Content
    slide = prs.slides.add_slide(slide_layout)
    slide.shapes.title.text = title
    slide.placeholders[1].text = content
```

```python
# Generate slides with LLM
for topic in topics:
    prompt = prompt_template.format_messages(topic=topic)
    response = llm.invoke(prompt)
    text = response.content

    # Simple logic to extract title + body (you can customize this better)
    lines = text.strip().split('\n', 1)
    title = lines[0].strip()
    content = lines[1].strip() if len(lines) > 1 else ""
    add_slide(title, content)
```

```python

# Save the presentation
prs.save("Class_11_Introduction_to_Accounting.pptx")
print("Presentation saved successfully.")
```

```python
## With Pydantic
from langchain_core.output_parsers import JsonOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
from pydantic import BaseModel, Field

model = ChatOpenAI(temperature=0.7)


# Define your desired data structure.
class Joke(BaseModel):
    setup: str = Field(description="question to set up a joke")
    punchline: str = Field(description="answer to resolve the joke")


# And a query intented to prompt a language model to populate the data structure.
joke_query = "Tell me a joke."

# Set up a parser + inject instructions into the prompt template.
parser = JsonOutputParser(pydantic_object=Joke)

prompt = PromptTemplate(
    template="Answer the user query.\n{format_instructions}\n{query}\n",
    input_variables=["query"],
    partial_variables={"format_instructions": parser.get_format_instructions()},
)

chain = prompt | model | parser

chain.invoke({"query": joke_query})
```

### Assisgment:
Create a simple assistant that uses any LLM and should be pydantic, when we ask about any product it should give you two information product Name, product details tentative price in USD (integer). use chat Prompt Template.


```python

from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain.prompts import ChatPromptTemplate
from langchain.output_parsers import PydanticOutputParser
from pydantic import BaseModel, Field

# Step 1: Define the output schema
class ProductInfo(BaseModel):
    product_name: str = Field(..., description="Name of the product")
    product_details: str = Field(..., description="Short description of the product")
    tentative_price_usd: int = Field(..., description="Tentative price in USD")
```

```python
# Step 2: Create the output parser
parser = PydanticOutputParser(pydantic_object=ProductInfo)
```

```python
# Step 3: Define the prompt template
prompt = PromptTemplate(
    template ="""You are a helpful product assistant. When given a product name, respond with:
- product name
- short product details
- a tentative price in USD (as an integer).\n

{format_instructions}\n

Product: {product_query}
""",
    input_variables=["product_query"],
    partial_variables={"format_instructions": parser.get_format_instructions()}
)
```

```python
# Step 4: Initialize the LLM (using GPT-3.5 or GPT-4 via OpenAI)
model = ChatOpenAI(model="gpt-3.5-turbo", temperature=0)
```

```python
# And a query intented to prompt a language model to populate the data structure.
product = "Apple iPhone 15 Pro"
```

```python
chain = prompt | model | parser
```

```python
result = chain.invoke({"product_query": product})
```
